**Situação**:

Uma equipe de operação de uma usina fotovoltaica deseja localizar períodos de alta geração e identificar quais inversores aparecem com maior frequência nesses momentos. Nesta atividade, o foco inicial será o arquivo de geração da planta, sem realizar junção com o arquivo de sensores.

2. Carregue a amostra e renomeie as principais variáveis para nomes mais claros, como Potencia_CC, Potencia_CA e Geracao_Diaria.

```python
df1.rename(columns={'DC_POWER': 'Potencia_CC',
                    'AC_POWER': 'Potencia_CA',
                    'DAILY_YIELD': 'Geracao_Diaria'
}, inplace=True)
```

```python
df1.columns
```

3.	Determine a maior potência CA registrada.




```python
maximo = df1['Potencia_CA'].max()
maximo
```

4.	Calcule um limiar correspondente a 70% da potência CA máxima.

```python
P_CA_70 = 0.70 * maximo
```

5.	Crie um DataFrame contendo somente registros acima desse limite e calcule sua quantidade e percentual.


```python
df_alto = df1[df1['Potencia_CA'] > P_CA_70]

quantidade = len(df_alto)

percentual = len(df_alto) / len(df1) * 100

print(quantidade, percentual)
```

6.	Utilize value_counts() em SOURCE_KEY dentro do DataFrame de alta geração.

```python
contagem_inversores = df_alto['SOURCE_KEY'].value_counts() # Um dataframe com value_counts() somente com registros SOURCE_KEY foi criado.
print(contagem_inversores)
```

7.	Identifique qual inversor aparece com maior frequência nos registros de alta geração e descreva o que esse resultado permite observar, sem concluir desempenho ou falha apenas com esse recorte.

```python
inversor_frequente = contagem_inversores.idxmax()
qtd_registros = contagem_inversores.max()
print(f"O inversor mais frequente é o '{inversor_frequente}' com {qtd_registros} registros.")
```

Este dado indica estritamente qual equipamento registrou o maior volume de amostras de tempo operando acima do limiar de 70% de potência. Mostra uma constância maior de operação em alta capacidade em relação aos demais.
